```python
import pandas as pd
import numpy as np
import os
from google.colab import files
from google.colab import drive
drive.mount('/content/drive')
```

Load Datasets

```python
colombo=pd.read_csv('/content/drive/MyDrive/Capstone_Datasets/Colombo.csv')
galle=pd.read_csv('/content/drive/MyDrive/Capstone_Datasets/Galle.csv')
gampaha=pd.read_csv('/content/drive/MyDrive/Capstone_Datasets/Gampaha.csv')
kandy=pd.read_csv('/content/drive/MyDrive/Capstone_Datasets/Kandy.csv')
jaffna=pd.read_csv('/content/drive/MyDrive/Capstone_Datasets/Jaffna.csv')
```

```python
print("Colombo :", colombo.shape)
print("Gampaha :", gampaha.shape)
print("Galle   :", galle.shape)
print("Kandy   :", kandy.shape)
print("Jaffna  :", jaffna.shape)
```

Basic Dataset Inspection

```python
datasets = {
    "Colombo": colombo,
    "Gampaha": gampaha,
    "Galle": galle,
    "Kandy": kandy,
    "Jaffna": jaffna
}

for district, df in datasets.items():

    print("\n" + "=" * 70)
    print(f"{district.upper()} DATASET")
    print("=" * 70)

    print("Shape:", df.shape)

    print("\nColumns:")
    print(df.columns.tolist())

    print("\nData types:")
    print(df.dtypes)

    print("\nFirst 5 rows:")
    display(df.head())
```

Standardize column names

```python
def clean_column_names(df):

    df = df.copy()

    df.columns = (
        df.columns
        .str.strip()
        .str.lower()
        .str.replace(" ", "_")
        .str.replace("-", "_")
        .str.replace("/", "_")
    )

    return df
```

```python
for district in datasets:

    datasets[district] = clean_column_names(
        datasets[district]
    )
```

```python
for district, df in datasets.items():

    print(f"\n{district}")
    print(df.columns.tolist())
```

Resetting the point IDs

```python
district_prefix = {
    "Colombo": "COL",
    "Gampaha": "GAM",
    "Galle": "GAL",
    "Kandy": "KAN",
    "Jaffna": "JAF"
}

for district, df in datasets.items():

    # Preserve original point ID
    df["pointid"] = df["pointid"].astype(int)

    # Create globally unique point ID
    df["global_pointid"] = (
        district_prefix[district]
        + "_"
        + df["pointid"].astype(str).str.zfill(3)
    )

    datasets[district] = df
```

```python
for district, df in datasets.items():

    print(f"\n{district}")

    display(
        df[
            [
                "district",
                "pointid",
                "global_pointid",
                "latitude",
                "longitude"
            ]
        ].drop_duplicates().head(5)
    )
```

Combining Five Datasets

```python
weather_df = pd.concat(
    datasets.values(),
    ignore_index=True
)

print("Combined dataset shape:")
print(weather_df.shape)

display(weather_df.head())
```
