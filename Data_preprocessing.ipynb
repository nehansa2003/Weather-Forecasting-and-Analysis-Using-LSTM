```python
import pandas as pd
import numpy as np
import os
from google.colab import files
from google.colab import drive
drive.mount('/content/drive')
```

Load Datasets

```python
colombo=pd.read_csv('/content/drive/MyDrive/Capstone_Datasets/Colombo.csv')
galle=pd.read_csv('/content/drive/MyDrive/Capstone_Datasets/Galle.csv')
gampaha=pd.read_csv('/content/drive/MyDrive/Capstone_Datasets/Gampaha.csv')
kandy=pd.read_csv('/content/drive/MyDrive/Capstone_Datasets/Kandy.csv')
jaffna=pd.read_csv('/content/drive/MyDrive/Capstone_Datasets/Jaffna.csv')
```

```python
print("Colombo :", colombo.shape)
print("Gampaha :", gampaha.shape)
print("Galle   :", galle.shape)
print("Kandy   :", kandy.shape)
print("Jaffna  :", jaffna.shape)
```

Basic Dataset Inspection

```python
datasets = {
    "Colombo": colombo,
    "Gampaha": gampaha,
    "Galle": galle,
    "Kandy": kandy,
    "Jaffna": jaffna
}

for district, df in datasets.items():

    print("\n" + "=" * 70)
    print(f"{district.upper()} DATASET")
    print("=" * 70)

    print("Shape:", df.shape)

    print("\nColumns:")
    print(df.columns.tolist())

    print("\nData types:")
    print(df.dtypes)

    print("\nFirst 5 rows:")
    display(df.head())
```

Standardize column names

```python
def clean_column_names(df):

    df = df.copy()

    df.columns = (
        df.columns
        .str.strip()
        .str.lower()
        .str.replace(" ", "_")
        .str.replace("-", "_")
        .str.replace("/", "_")
    )

    return df
```

```python
for district in datasets:

    datasets[district] = clean_column_names(
        datasets[district]
    )
```

```python
for district, df in datasets.items():

    print(f"\n{district}")
    print(df.columns.tolist())
```

Resetting the point IDs

```python
district_prefix = {
    "Colombo": "COL",
    "Gampaha": "GAM",
    "Galle": "GAL",
    "Kandy": "KAN",
    "Jaffna": "JAF"
}

for district, df in datasets.items():

    # Preserve original point ID
    df["pointid"] = df["pointid"].astype(int)

    # Create globally unique point ID
    df["global_pointid"] = (
        district_prefix[district]
        + "_"
        + df["pointid"].astype(str).str.zfill(3)
    )

    datasets[district] = df
```

```python
for district, df in datasets.items():

    print(f"\n{district}")

    display(
        df[
            [
                "district",
                "pointid",
                "global_pointid",
                "latitude",
                "longitude"
            ]
        ].drop_duplicates().head(5)
    )
```

Combining Five Datasets

```python
weather_df = pd.concat(
    datasets.values(),
    ignore_index=True
)

print("Combined dataset shape:")
print(weather_df.shape)

display(weather_df.head())
```

```python
print(
    weather_df["district"].value_counts()
)
```

```python
# CHECK GLOBAL POINT ID UNIQUENESS


total_points = weather_df["global_pointid"].nunique()

print("Total unique geographical points:", total_points)
```

```python
# Check points per district

points_per_district = (
    weather_df
    .groupby("district")["global_pointid"]
    .nunique()
)

print(points_per_district)
```

Verify Point Location Consistency

```python
point_locations = (
    weather_df
    .groupby("global_pointid")
    .agg(
        district=("district", "nunique"),
        latitude=("latitude", "nunique"),
        longitude=("longitude", "nunique")
    )
    .reset_index()
)

problematic_points = point_locations[
    (point_locations["district"] > 1) |
    (point_locations["latitude"] > 1) |
    (point_locations["longitude"] > 1)
]

print(
    "Points with inconsistent locations:",
    len(problematic_points)
)

display(problematic_points.head())
```

```python
coordinate_check = (
    weather_df
    .groupby("global_pointid")
    .agg(
        latitude_count=("latitude", "nunique"),
        longitude_count=("longitude", "nunique"),
        elevation_count=("elevation_m", "nunique")
    )
    .reset_index()
)

problem_points = coordinate_check[
    (coordinate_check["latitude_count"] > 1) |
    (coordinate_check["longitude_count"] > 1)
]

print(
    "Points with inconsistent coordinates:",
    len(problem_points)
)
```

Record Count Per Point

```python
records_per_point = (
    weather_df
    .groupby(
        ["district", "global_pointid"]
    )
    .size()
    .reset_index(
        name="record_count"
    )
)
display(
    records_per_point
    .groupby("district")["record_count"]
    .describe()
)
```

Check Missing values

Check hourly continuity

```python
weather_df = weather_df.sort_values(
    ["district", "pointid", "datetime"]
)

weather_df["time_difference"] = (
    weather_df
    .groupby(["district", "pointid"])["datetime"]
    .diff()
)
```

```python
gaps = weather_df[
    weather_df["time_difference"] > pd.Timedelta(hours=1)
]

print("Number of gaps:", len(gaps))

display(
    gaps[
        [
            "district",
            "pointid",
            "datetime",
            "time_difference"
        ]
    ].head(20)
)
```

Time-based features

```python
weather_df["hour"] = weather_df["datetime"].dt.hour
weather_df["day_of_week"] = weather_df["datetime"].dt.dayofweek
weather_df["day_of_year"] = weather_df["datetime"].dt.dayofyear

weather_df["is_weekend"] = (
    weather_df["day_of_week"] >= 5
).astype(int)
```

Cyclical Encoding

```python
weather_df["hour_sin"] = np.sin(
    2 * np.pi * weather_df["hour"] / 24
)

weather_df["hour_cos"] = np.cos(
    2 * np.pi * weather_df["hour"] / 24
)

weather_df["month_sin"] = np.sin(
    2 * np.pi * weather_df["month"] / 12
)

weather_df["month_cos"] = np.cos(
    2 * np.pi * weather_df["month"] / 12
)
```

```python
cyclical_columns = [
    "hour_sin",
    "hour_cos",
    "month_sin",
    "month_cos"
]

print("Checking cyclical feature columns:\n")

for column in cyclical_columns:

    if column in weather_df.columns:
        print(f"{column} exists")
    else:
        print(f"{column} NOT FOUND")
```

```python
# COMPARE RAINFALL COLUMNS
print(
    weather_df[
        ["rainfall_mm", "rain_mm"]
    ].describe()
)

print("\nCorrelation:")
print(
    weather_df[
        ["rainfall_mm", "rain_mm"]
    ].corr()
)
```

```python
difference = (
    weather_df["rainfall_mm"]
    - weather_df["rain_mm"]
).abs()

print(
    "Number of different values:",
    (difference > 0.0001).sum()
)

print(
    "Maximum difference:",
    difference.max()
)
```

Removing Unecessary Columns

```python
# rain_mm is exactly identical to rainfall_mm
weather_df = weather_df.drop(
    columns=["rain_mm"]
)

# visibility_m contains only missing values
weather_df = weather_df.drop(
    columns=["visibility_m"]
)

print("Removed columns:")
print("- rain_mm       → duplicate of rainfall_mm")
print("- visibility_m  → 100% missing")

print("\nRemaining columns:")
print(weather_df.columns.tolist())
```

Physical Validity Checking

```python
# Humidity
print(
    "Humidity < 0:",
    (weather_df["humidity_pct"] < 0).sum()
)

print(
    "Humidity > 100:",
    (weather_df["humidity_pct"] > 100).sum()
)

# Wind speed
print(
    "Negative wind speed:",
    (weather_df["windspeed_kmh"] < 0).sum()
)

# Rainfall
print(
    "Negative rainfall_mm:",
    (weather_df["rainfall_mm"] < 0).sum()
)

# Cloud cover
print(
    "Cloud cover < 0:",
    (weather_df["cloudcover_pct"] < 0).sum()
)

print(
    "Cloud cover > 100:",
    (weather_df["cloudcover_pct"] > 100).sum()
)
```

Organizing Columns

```python
column_order = [
    # Identification
    "district",
    "pointid",
    "global_pointid",

    # Geography
    "latitude",
    "longitude",
    "elevation_m",

    # Time
    "datetime",
    "date",
    "day",
    "month",
    "year",
    "time",
    "season",

    # Weather
    "temperature_c",
    "feelslike_c",
    "humidity_pct",
    "dewpoint_c",
    "pressure_hpa",
    "windspeed_kmh",
    "winddirection_deg",
    "winddirection_compass",
    "windgust_kmh",
    "rainfall_mm",
    "cloudcover_pct",
    "solarradiation_wm2",
    "sunshineduration_s",
    "visibility_m"
]

# Keep only columns that actually exist
column_order = [
    col for col in column_order
    if col in weather_df.columns
]

weather_df = weather_df[column_order].copy()

print("Final columns:")
print(weather_df.columns.tolist())
```

Save Processed Data

```python
PROCESSED_DIR = "/content/drive/MyDrive/Capstone_Datasets"

WEATHER_FILE = os.path.join(
    PROCESSED_DIR,
    "weather_data_processed_2020_2025.parquet"
)

weather_df.to_parquet(
    WEATHER_FILE,
    index=False,
    engine="pyarrow"
)

print("Processed dataset saved successfully.")

print("\nFile:")
print(WEATHER_FILE)

print("\nDataset shape:")
print(weather_df.shape)

print("\nFile size:")
file_size_mb = os.path.getsize(WEATHER_FILE) / (1024 ** 2)
print(f"{file_size_mb:.2f} MB")
```

```python

# Save Geographical Point References
point_reference = (
    weather_df[
        [
            "district",
            "pointid",
            "global_pointid",
            "latitude",
            "longitude",
            "elevation_m"
        ]
    ]
    .drop_duplicates()
    .sort_values(
        ["district", "pointid"]
    )
    .reset_index(drop=True)
)

POINT_FILE = os.path.join(
    PROCESSED_DIR,
    "weather_point_reference.csv"
)

point_reference.to_csv(
    POINT_FILE,
    index=False
)


print("Geographycal Point References Saved.")

print("\nFile:")
print(POINT_FILE)

print("\nNumber of geographical points:")
print(len(point_reference))

display(point_reference.head(10))
```
