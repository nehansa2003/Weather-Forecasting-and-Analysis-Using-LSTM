Basic Dataset Inspection

```python
datasets = {
    "Colombo": colombo,
    "Gampaha": gampaha,
    "Galle": galle,
    "Kandy": kandy,
    "Jaffna": jaffna
}

for district, df in datasets.items():

    print("\n" + "=" * 70)
    print(f"{district.upper()} DATASET")
    print("=" * 70)

    print("Shape:", df.shape)

    print("\nColumns:")
    print(df.columns.tolist())

    print("\nData types:")
    print(df.dtypes)

    print("\nFirst 5 rows:")
    display(df.head())
```
